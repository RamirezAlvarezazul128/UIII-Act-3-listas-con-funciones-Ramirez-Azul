#### Listas de modelo

```python
# =========================
# MENÚ DE LISTAS (GUARDIAS)
# =========================

lista_guardias = []

# =========================
# Agregar
# =========================
def agregarGuardia():
    nombre = input("Ingrese nombre del guardia: ")
    rango = input("Ingrese rango (ej. Supervisor, Operativo): ")
    turno = input("Ingrese turno (ej. Mañana, Tarde, Noche): ")

    guardia = {
        "nombre": nombre,
        "rango": rango,
        "turno": turno
    }

    lista_guardias.append(guardia)
    print("Guardia registrado correctamente.\n")


# =========================
# Leer
# =========================
def leerGuardia():

    if len(lista_guardias) == 0:
        print("No hay guardias registrados.\n")
        return

    posicion = int(input("Ingrese posición del guardia en la lista: "))

    if 0 <= posicion < len(lista_guardias):
        print(lista_guardias[posicion])
    else:
        print("Posición inválida.\n")


# =========================
# Listar (TABLA ALINEADA)
# =========================
def listarGuardias():

    if len(lista_guardias) == 0:
        print("Lista de guardias vacía.\n")
        return

    print("\n======================= REPORTE DE GUARDIAS =======================")

    # Encabezado con anchos fijos: Posición (5), Nombre (25), Rango (15), Turno (12)
    print(f"{'Pos.':<5} | {'Nombre':<25} | {'Rango':<15} | {'Turno':<12}")
    print("-" * 67) # Línea divisoria

    # Filas con el mismo ancho exacto que el encabezado
    for i, guardia in enumerate(lista_guardias):
        print(f"{i:<5} | {guardia['nombre']:<25} | {guardia['rango']:<15} | {guardia['turno']:<12}")

    print("===================================================================\n")


# =========================
# Buscar
# =========================
def buscarGuardia():

    nombre = input("Ingrese nombre del guardia a buscar: ")

    encontrado = False

    for g in lista_guardias:

        if g["nombre"].lower() == nombre.lower():
            print("Guardia encontrado:")
            print(g)
            encontrado = True

    if not encontrado:
        print("Guardia no encontrado.\n")


# =========================
# Modificar
# =========================
def modificarGuardia():

    nombre = input("Ingrese nombre del guardia a modificar: ")

    for g in lista_guardias:

        if g["nombre"].lower() == nombre.lower():

            nuevoNombre = input("Nuevo nombre: ")
            nuevoRango = input("Nuevo rango: ")
            nuevoTurno = input("Nuevo turno: ")

            g["nombre"] = nuevoNombre
            g["rango"] = nuevoRango
            g["turno"] = nuevoTurno

            print("Datos del guardia modificados correctamente.\n")
            return

    print("Guardia no encontrado.\n")


# =========================
# Borrar
# =========================
def borrarGuardia():

    nombre = input("Ingrese nombre del guardia a dar de baja: ")

    for g in lista_guardias:

        if g["nombre"].lower() == nombre.lower():

            lista_guardias.remove(g)

            print("Guardia eliminado del sistema correctamente.\n")
            return

    print("Guardia no encontrado.\n")


# =========================
# MENÚ PRINCIPAL
# =========================

while True:

    print("===== MENÚ DE CONTROL DE GUARDIAS =====")
    print("1. Registrar guardia")
    print("2. Leer datos de un guardia (por posición)")
    print("3. Listar todos los guardias")
    print("4. Buscar guardia por nombre")
    print("5. Modificar datos de un guardia")
    print("6. Dar de baja (borrar) guardia")
    print("7. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregarGuardia()

    elif opcion == "2":
        leerGuardia()

    elif opcion == "3":
        listarGuardias()

    elif opcion == "4":
        buscarGuardia()

    elif opcion == "5":
        modificarGuardia()

    elif opcion == "6":
        borrarGuardia()

    elif opcion == "7":
        print("Saliendo del sistema de guardias...")
        break

    else:
        print("Opción inválida.\n")
```
